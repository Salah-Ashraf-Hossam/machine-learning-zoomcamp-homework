```python
df=pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/course_lead_scoring_2026.csv')
```

```python
categorical=['lead_source','industry','employment_status','location']
```

```python
numerical=['annual_income','number_of_courses_viewed','interaction_count','lead_score','converted']
```

```python
df[categorical]=df[categorical].fillna('nan')
```

```python
df[numerical]=df[numerical].fillna(0)
```

```python
df.isnull().sum()
```

```python
df.industry.value_counts()
```

```python
for c in numerical :
    for b in numerical:
        if c != b :
            corr=round(df[c].corr(df[b]),3)
            print(f"the correlation of  {c}  with   {b}   equals   =   {corr}")
```

```python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=42)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=42)
```

```python
y_train = df_train.converted.values
y_val = df_val.converted.values
y_test = df_test.converted.values

del df_train['converted']
del df_val['converted']
del df_test['converted']
```

```python
y_full_train = df_full_train.converted.values
del df_full_train['converted']
```

```python
def mutual_info_converted_score(series):
    return mutual_info_score(series, y_train)
```

```python
mi = round (df_train[categorical].apply(mutual_info_converted_score),2)
mi.sort_values(ascending=False)
```
