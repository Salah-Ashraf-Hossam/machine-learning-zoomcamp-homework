```python
Asia_df=df[df['origin']=='Asia'][['vehicle_weight','model_year']].head(7)
Asia_df
```

```python
x=np.array(Asia_df)
x
```

```python
XTX=x.T.dot(x)
XTX
```

```python
XTX_inv=np.linalg.inv(XTX)
XTX_inv
```

```python
y=np.array([1100,1300,800,900,1000,1100,1200])
y
```

```python
w=(XTX_inv.dot(x.T)).dot(y)
w
```
